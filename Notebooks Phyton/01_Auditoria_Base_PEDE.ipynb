```python
def resumo_nulos(df, nome):

    resumo = (
        pd.DataFrame({
            'coluna': df.columns,
            'nulos': df.isnull().sum(),
            'percentual': round((df.isnull().sum() / len(df)) * 100, 2)
        })
        .sort_values('percentual', ascending=False)
    )

    print(f"\n{'='*20} {nome} {'='*20}")
    display(resumo.head(15))
```

```python
resumo_nulos(df2022, '2022')
resumo_nulos(df2023, '2023')
resumo_nulos(df2024, '2024')
```

```python
ra2022 = set(df2022['RA'])
ra2023 = set(df2023['RA'])
ra2024 = set(df2024['RA'])
```

```python
print("RA em 2022:", len(ra2022))
print("RA em 2023:", len(ra2023))
print("RA em 2024:", len(ra2024))
```

```python
print("2022 e 2023:", len(ra2022.intersection(ra2023)))

print("2023 e 2024:", len(ra2023.intersection(ra2024)))

print("2022, 2023 e 2024:",
      len(ra2022.intersection(ra2023).intersection(ra2024)))
```

```python
print("Novos alunos em 2023:",
      len(ra2023 - ra2022))

print("Novos alunos em 2024:",
      len(ra2024 - ra2023))
```
