# Resumo Executivo

## Principais Descobertas

- 46% dos estudantes apresentam algum grau de defasagem.
- Engajamento é um dos fatores mais associados ao desempenho acadêmico.
- Aprendizagem, Engajamento e Ponto de Virada são os indicadores mais associados ao INDE.
- Estudantes com menor INDE inicial apresentaram a maior evolução ao longo do programa.
- 79,6% dos alunos mantiveram ou evoluíram sua classificação de Pedra.
- O modelo preditivo atingiu 74% de acurácia e identificou 81% dos alunos em risco futuro de defasagem.

# Slide 3 - Defasagem Educacional

## Principais Resultados

- 53,8% dos alunos apresentaram IAN = 10 (adequação ao nível).
- 45,9% apresentaram IAN = 5.
- 0,3% apresentaram IAN = 2,5.

## Insight

Aproximadamente 46% dos estudantes apresentam algum grau de defasagem em relação à fase ideal.

Esse resultado reforça a importância de mecanismos de acompanhamento e intervenção precoce.

# Slide 5 - O Engajamento Como Motor da Aprendizagem

## Evidência

| Faixa de Engajamento | IDA Médio |
|------|------:|
| Baixo Engajamento | 3,92 |
| Engajamento Médio | 5,44 |
| Alto Engajamento | 7,12 |

## Insight

O engajamento apresentou forte associação com a aprendizagem.

Alunos altamente engajados apresentaram desempenho acadêmico médio significativamente superior ao observado entre alunos com baixo engajamento.

## Mensagem-Chave

Engajamento não é consequência da aprendizagem. Os dados sugerem que ele é um dos principais motores do desempenho acadêmico.

# Slide 9 - Modelo Preditivo de Risco de Defasagem

## Desempenho

- Accuracy: 74,16%
- Recall: 81%
- Precision: 73%
- F1-Score: 76%

## Variáveis Mais Importantes

1. IPV
2. IDA
3. IEG

## Insight

O modelo foi capaz de identificar 81% dos estudantes que efetivamente entrariam em risco de defasagem.

## Mensagem-Chave

É possível utilizar os indicadores atuais dos estudantes para criar mecanismos de alerta precoce.

# Slide 10 - Recomendações Estratégicas

## Recomendação 1

Priorizar ações que aumentem o engajamento dos estudantes.

## Recomendação 2

Monitorar continuamente o Indicador de Ponto de Virada (IPV).

## Recomendação 3

Utilizar o modelo preditivo como ferramenta de identificação precoce de risco.

## Recomendação 4

Concentrar esforços nos estudantes com menor INDE inicial, grupo que apresentou maior potencial de evolução.

## Conclusão Final

Os dados indicam que o desenvolvimento educacional dos estudantes está fortemente associado à aprendizagem, engajamento e ponto de virada. Além disso, há evidências de efetividade do programa e viabilidade de utilização de modelos preditivos para apoio à tomada de decisão.
